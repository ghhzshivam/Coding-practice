#### LIST



A list is an ordered, mutable, collection of elements.

Elements can be of any data type (int, str, float, list, etc.).

Lists are defined using square brackets [].


Lists are mutable, tuples are immutable.

Indexing can be negative.

Slicing returns a new list, original is unchanged.

List comprehension is faster than loops.

append() adds one element, extend() adds iterable elements.

copy() is shallow → nested lists can still be linked.

sort() vs sorted() difference: in-place vs new list.

```python
my_list = [1, 2, 3, 4]
mixed_list = [1, "hello", 3.5, [1,2]]
empty_list = []
lst = [10, 20, 30, 40]
print(lst[0])  # 10
print(lst[-1]) # 40 (last element)
lst[1] = 99  # [1,99,3]

```

```python
# 3.1 Adding Elements
# append(x) → adds single element at the end
# extend(iterable) → adds all elements of iterable
# insert(index, x) → inserts element at specific index

lst = [1,2,3]
lst.append(4)       # [1,2,3,4]
lst.extend([5,6])   # [1,2,3,4,5,6]
lst.insert(2, 99)   # [1,2,99,3,4,5,6]
```

```python
# 3.2 Removing Elements
# remove(x) → remove first occurrence of x
# pop([index]) → remove element at index (default last)
# del lst[index] → remove element at index
# clear() → remove all elements

lst = [1,2,3,2,4]
lst.remove(2)   # [1,3,2,4]
lst.pop()       # [1,3,2] (removes last)
del lst[1]      # [1,2]
lst.clear()     # []
```

```python
# List Queries & Information
# len(lst) → length
# min(lst) / max(lst) → min/max value
# sum(lst) → sum of elements (numbers only)
# count(x) → count occurrences
# index(x, start, end) → find first index
# in / not in → check presence

lst = [1,2,3,2]
print(len(lst))     # 4
print(lst.count(2)) # 2
print(2 in lst)     # True
print(lst.index(3)) # 2
```

```python
# Sorting & Reversing
# sort() → sorts ascending (in-place)
# sort(reverse=True) → descending
# sorted(lst) → returns sorted copy
# reverse() → reverses in-place
# lst[::-1] → reversed copy

lst = [3,1,4,2]
lst.sort()          # [1,2,3,4]
lst.sort(reverse=True)  # [4,3,2,1]
print(sorted(lst))  # returns [1,2,3,4]
lst.reverse()       # reverse in-place
```

#### Dict

A dictionary is an unordered (Python <3.7), mutable collection of key-value pairs.

Keys must be immutable (int, str, tuple), values can be any type (int, str, list, dict, etc.).

Dictionaries are defined using curly braces {}:

Access elements: d[key] or d.get(key, default) → avoids KeyError.

Mutable → keys can’t change, values can be updated.

Add/Update: d[key] = value or d.update({'key': value}).

Delete: del d[key], d.pop(key), d.popitem() (last inserted key).

Copying:

copy() → shallow (nested dicts still linked)

deepcopy() → fully independent copy

Iterating:

d.keys() → keys

d.values() → values

d.items() → key-value pairs

Merge: d1.update(d2) (Python <3.9), d3 = d1 | d2 (Python 3.9+)

Sorting:

By key: dict(sorted(d.items()))

By value: dict(sorted(d.items(), key=lambda x: x[1]))

reverse=True → descending order

Nested dictionaries: Access with d[key1][key2].

Tricks/Useful methods:

setdefault(key, default) → create key if missing

max(d, key=d.get) → max value key

min(d, key=d.get) → min value key

Swap keys/values: {v:k for k,v in d.items()}

Important points:

d.items(), d.keys(), d.values() are views, not lists

Dictionary preserves insertion order from Python 3.7+

```python
d = {}

# With values
d = {'a': 1, 'b': 2, 'c': 3}
# Using dict() constructor
d = dict(a=1, b=2, c=3)
# From a list of tuples
d = dict([('a', 1), ('b', 2)])
# From keys with same value
d = dict.fromkeys(['x', 'y', 'z'], 0)  # {'x':0, 'y':0, 'z':0}


keys = ['a', 'b', 'c']
values = [1, 2, 3]

# Using zip
d = dict(zip(keys, values))
print(d)  # {'a': 1, 'b': 2, 'c': 3}   #If lists are unequal, zip stops at the shortest list.
```

```python
d = {'a': 1, 'b': 2}

# Access by key
print(d['a'])  # 1

# Access safely using get()
print(d.get('c'))         # None
print(d.get('c', 0))      # 0 (default value if key not present)

d.update({'b': 20, 'd': 4, 'a': 2}) # if new dict have key which is in old dict it will update old
d
```

```python
# Deleting Elements
d = {'a': 1, 'b': 2, 'c': 3}

# Delete key
del d['b']        # {'a':1, 'c':3}

# Remove and return value
val = d.pop('c')  # val=3, d={'a':1}

# Remove last inserted item (Python 3.7+)
d.popitem()        # removes last key-value pair

# Clear dictionary
d.clear()          
```

```python
# Special Dictionary Types (Advanced, Sometimes Asked)

# defaultdict (from collections) → avoids KeyError:

from collections import defaultdict
d = defaultdict(int)
d['a'] += 1  # default 0


# OrderedDict (Python <3.7 → maintains order):

from collections import OrderedDict
od = OrderedDict()
od['a'] = 1
od['b'] = 2


# Counter → counts elements:

from collections import Counter
c = Counter(['a','b','a'])
print(c)  # Counter({'a':2,'b':1})
```

```python
# Sorting a Dictionary

# You cannot sort a dict in place (until Python 3.7+, order is preserved), but you can get sorted lists of items or create a new sorted dictionary.

# a) Sort by Key
d = {'b': 2, 'a': 3, 'c': 1}

# Ascending by key
d_sorted_asc = dict(sorted(d.items()))
print(d_sorted_asc)  # {'a': 3, 'b': 2, 'c': 1}

# Descending by key
d_sorted_desc = dict(sorted(d.items(), reverse=True))
print(d_sorted_desc)  # {'c': 1, 'b': 2, 'a': 3}

# b) Sort by Value
d = {'b': 2, 'a': 3, 'c': 1}

# Ascending by value
d_val_asc = dict(sorted(d.items(), key=lambda item: item[1]))
print(d_val_asc)  # {'c': 1, 'b': 2, 'a': 3}

# Descending by value
d_val_desc = dict(sorted(d.items(), key=lambda item: item[1], reverse=True))
print(d_val_desc)  # {'a': 3, 'b': 2, 'c': 1}
```

#### Tuple

A tuple is an ordered, immutable collection of items, which can contain elements of different data types.

Tuples use less memory than lists; often used for immutable records.

Single element tuple must have a comma.

Tuples are immutable, but lists inside tuples are mutable.

t = (1,2,3) * 2 → (1,2,3,1,2,3)

t[1:3] → returns a tuple, not list.

Tuples can be dictionary keys (immutable).

No tuple comprehension: (x for x in range(5)) is generator.

Tuple comparison is lexicographical.

tuple() can convert iterables to tuples.

Only count() and index() are valid tuple methods

```python
t1 = (1, 2, 3)
t = 1, 2, 3

t2 = ("apple", "banana", "cherry")
t3 = (1, "apple", 3.14)
empty = ()

# Single-element tuple:
# Tricky: A single element requires a comma ,

single = (5,)   # tuple
not_tuple = (5) # int
```

```python
t = (23, 45, 67, 23, 89, 45, 78)
t.index(45, 5)

tuple([2])
```

```python
# Accessing Tuple Elements

# Indexing:

t = (10, 20, 30)
print(t[0])   # 10
print(t[-1])  # 30


# Slicing:

t = (0,1,2,3,4,5)
print(t[1:4])   # (1,2,3)
print(t[:3])    # (0,1,2)
print(t[3:])    # (3,4,5)
print(t[::-1])  # reversed tuple
```

```python
# Tuple Functions / Methods

# Tuples have very few methods because they are immutable. Important ones:

# count(x) → Returns how many times x appears

t = (1,2,2,3)
print(t.count(2))  # 2


# index(x) → Returns first index of x

t = (10, 20, 30)
print(t.index(20))  # 1


# Note: Tuples do not have append, remove, pop, extend, or sort methods (immutable).
```

```python
# Tuple Packing & Unpacking

# Packing: Assign multiple values to a tuple

t = 1, 2, 3   # packed tuple


# Unpacking: Assign tuple elements to variables

a, b, c = t
print(a, b, c)  # 1 2 3


# Extended unpacking (Python 3+):

t = (1,2,3,4,5)
a, *b, c = t
print(a, b, c)  # 1 [2,3,4
```
